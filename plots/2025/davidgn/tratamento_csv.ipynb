```python
import plotly.express as px
import pandas as pd
from tratamento_csv import  combinar_csvs_em_dicio, expand_dicts, indices_maiores_que
import plotly.graph_objects as go
import tratamento_csv as tc
path_comum="../../../results/results_flows"

path_msf = path_comum+"/msf_s_50_p_4_a_1.0_c_0"
path_kuririn = path_comum + "/kuririn_ppo_s_50_p_4_a_1.0_c_0"


# path_vegeta = "/home/lacis/Área de trabalho/teste/multi-user-sfc/results/results_flows/vegeta_s_50_p_4_a_1.0_c_0/2025-04-30_09-50-02_id7419.csv"
dicio_csv_kuririn = combinar_csvs_em_dicio(path_kuririn)
# dicio_csv_GA = combinar_csvs_em_dicio(path_GA)
# dicio_csv_MSF = combinar_csvs_em_dicio(path_msf)
# dicio_csv_osfem = combinar_csvs_em_dicio(path_osfem)
# dicio_csv_goku = combinar_csvs_em_dicio(path_goku)

# dicio_csv_vegeta = extrair_dicio_csv(path_vegeta)

# dicio_csv_kuririn, dicio_csv_GA,dicio_csv_MSF,dicio_csv_osfem, dicio_csv_goku=expand_dicts(dicio_csv_kuririn,dicio_csv_GA,dicio_csv_MSF,dicio_csv_osfem,dicio_csv_goku)

dicio_csv_kuririn=expand_dicts(dicio_csv_kuririn)[0]
print(dicio_csv_kuririn)

# lista_outliers = indices_maiores_que(dicio_csv_kuririn['decision_time_ms'], 100)
# for i in lista_outliers:
#     dicio_csv_kuririn['decision_time_ms'][i] = 100
tempos = []
aux = [200, 400, 600, 800, 1000]
i = 0
for tempo in dicio_csv_kuririn['time_seconds']:

    if i >= len(aux):
        tempos.append(aux[-1])

    elif float(tempo) < aux[i]:
        tempos.append(aux[i])
    else:
        i += 1
        tempos.append(tempos[i])
```

```python
data = {
    'Time (s)': tempos,
    'Kuririn (PPO)': dicio_csv_kuririn["acceptance_rate"],
    # 'GA': dicio_csv_GA["acceptance_rate"],
    # 'MSF': dicio_csv_MSF["acceptance_rate"],
    # 'OSFEM': dicio_csv_osfem["acceptance_rate"],
    # 'Goku': dicio_csv_goku["acceptance_rate"]
    # 'vegeta': dicio_csv_vegeta["acceptance_rate"]
}

df = pd.DataFrame(data)

df_long = df.melt(id_vars=['Time (s)'], var_name='Method', value_name='Acceptance Rate (%)')
```

```python
# Criando o gráfico de box plot
fig = px.box(df_long, x='Time (s)', y='Acceptance Rate (%)', color='Method', 
             title=None,orientation='v', points=False)

fig.update_traces(quartilemethod="exclusive") 


draft_template = go.layout.Template()
draft_template.layout.annotations = [
    dict(
        name="draft watermark",
        text="",
        textangle=-30,
        opacity=0.1,
        font=dict(color="black", size=100),
        xref="paper",
        yref="paper",
        x=0.5,
        y=0.5,
        showarrow=False,
    )
]

fig.update_layout(
    yaxis=dict(showline=True,showgrid=True),
    xaxis=dict(showline=True,showgrid=True),
    legend_title=None,
    margin=dict(l=100, r=10, b=80, t=25),
    autosize=True,
    width=950,
    height=600,
    template=draft_template,
    legend=dict(x=0.05, y=1.2, traceorder='normal', orientation='h'),
    font=dict(family="Arial",  size=25, color="Black")
)

# Exibindo o gráfico
fig.show()
```
